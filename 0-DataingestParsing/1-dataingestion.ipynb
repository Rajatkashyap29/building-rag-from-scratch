### Directory Loader 


```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

# Load all .txt files
dir_loader = DirectoryLoader(
    "Data/text_files",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=True
)

# Load documents
documents = dir_loader.load()

# Total documents
print("=" * 60)
print(f"📄 Total Documents Loaded: {len(documents)}")
print("=" * 60)

# Print each document
for i, doc in enumerate(documents, start=1):
    print(f"\n📑 Document {i}")
    print("-" * 60)

    print("📂 File Path:")
    print(doc.metadata.get("source"))

    print("\n📝 Content Preview:")
    print(doc.page_content[:300])  # First 300 characters

    print("\n📌 Metadata:")
    print(doc.metadata)

    print("=" * 60)
```

### method -1 character text splitter 

```python
text = documents[0].page_content
text
```

```python
#character text splitter

char_splitter = CharacterTextSplitter(
    separator='/n',
    chunk_size = 200,
    chunk_overlap = 20,
    length_function = len
)

char_chunk = char_splitter.split_text(text)
print("Total chunk is -> ",len(char_chunk))
print(char_chunk[0][:100])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

recursive_char_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""],
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)

recursive_chunks = recursive_char_splitter.split_text(text)


print(f"📄 Total Chunks: {len(recursive_chunks)}")


for i, chunk in enumerate(recursive_chunks, start=1):
    print(f"\n📌 Chunk {i}")
    print("-" * 70)
    print(f"📏 Length: {len(chunk)} characters\n")
    print(chunk)
    print("=" * 70)
```

```python
#Token based splitting 

token_splitter = TokenTextSplitter(
    chunk_size = 50 ,
    chunk_overlap = 10
    
)

token_chunks = token_splitter.split_text(text)
print(f"📄 Total Chunks: {len(token_chunks)}")
print(f"first chuks:",token_chunks[0][:100])

```
