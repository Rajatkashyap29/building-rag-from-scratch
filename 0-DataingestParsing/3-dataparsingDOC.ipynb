### word Document Parsing

```python
from langchain_community.document_loaders import Docx2txtLoader,UnstructuredWordDocumentLoader
```

```python
##method -1 docx2loader
try:
    docx_loader = Docx2txtLoader("Data/word-file/Prolog_Lab_Manual.docx")
    docs = docx_loader.load()
    print("DOCX LOADED ")
    print(f"TOTAL DOCS {len(docs)}")
    print(f"CONTENT:\n {docs[0].page_content[:200]}")
    print(f"METADATA:\n {docs[0].metadata}")
    
except Exception as e :
    print(f"Error {e}")    
```

```python
##method - 2 UNSTRUCTURED doc loader
try:
    unstructured_loader = UnstructuredWordDocumentLoader("Data/word-file/Prolog_Lab_Manual.docx",mode="elements")
    doc = unstructured_loader.load()
    print("DOCX LOADED ")
    print(f"TOTAL DOCS {len(doc)}")
    print(f"CONTENT:\n {doc[0].page_content[:200]}")
    print(f"METADATA:\n {doc[0].metadata}")
    
except Exception as e :
    print(f"Error {e}")        
```
