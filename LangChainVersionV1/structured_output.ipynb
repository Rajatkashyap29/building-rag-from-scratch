## Structured Output
### . Structured Output is a technique for making an LLM return its response in a predefined structure or schema, such as JSON or a Pydantic model.

## Using Pydantic

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
import os

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
llm = init_chat_model(
    model="groq:openai/gpt-oss-120b"
)
llm
```

```python
from pydantic import BaseModel,Field
class Movie(BaseModel):
    title:str = Field(description="The Title of The Movie")
    year:int = Field(deprecated="Year of The Movie Released")
    director:str = Field(description="The Director Of The Movie")
    rating:bool = Field(description="The Rating of The Movie")
```

```python
structured_output = llm.with_structured_output(Movie)
structured_output
```

```python
res = llm.invoke("Give me detail about Movie Wanted")
res
```

```python
response = structured_output.invoke("Provide Detail about movie Reacher")
response
```

## TypeDict

```python
from typing_extensions import TypedDict

class Student(TypedDict):
    name: str
    age: int
    skill: str

structured_llm = llm.with_structured_output(Student)

response = structured_llm.invoke(
    "My name is Rajat, I am 21 years old and I know Python."
)

print(response)
```
