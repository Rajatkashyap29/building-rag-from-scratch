```python
from langchain_huggingface import HuggingFaceEmbeddings
embedding = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-miniLM-L6-V2"
)
embedding
```

```python
import os
from dotenv import load_dotenv
from astrapy import DataAPIClient

load_dotenv()

endpoint = os.getenv("ASTRA_DB_API_ENDPOINT")
token = os.getenv("ASTRA_DB_APPLICATION_TOKEN")

client = DataAPIClient(token)

db = client.get_database(
    endpoint,
    keyspace="rag_keyspace"
)

print("Astra connection successful!")
```

```python
from langchain_astradb import AstraDBVectorStore

vectorstore = AstraDBVectorStore(
    embedding=embedding,
    collection_name="rag_collection",
    api_endpoint=endpoint,
    token=token,
    namespace="rag_keyspace"
)

print("Astra Vector Store created!")
vectorstore
```

```python
from langchain_core.documents import Document

sample_documents = [
    Document(
        page_content="""
Artificial Intelligence (AI) is a field of computer science that focuses on
creating systems capable of performing tasks that normally require human
intelligence. These tasks include reasoning, learning, problem solving,
decision making, speech recognition, and understanding natural language.

AI is used in many real-world applications such as recommendation systems,
chatbots, autonomous vehicles, fraud detection, and medical diagnosis.
        """,
        metadata={
            "source": "AI Introduction",
            "topic": "AI",
            "page": 1
        }
    ),

    Document(
        page_content="""
Machine Learning (ML) is a subset of Artificial Intelligence in which
computers learn patterns from data instead of being explicitly programmed
for every task.

The main types of machine learning are supervised learning, unsupervised
learning, and reinforcement learning. Supervised learning uses labelled data
for tasks such as classification and regression. Unsupervised learning finds
hidden patterns in unlabelled data, while reinforcement learning learns
through rewards and penalties.
        """,
        metadata={
            "source": "ML Basics",
            "topic": "ML",
            "page": 1
        }
    ),

    Document(
        page_content="""
Natural Language Processing (NLP) is a branch of Artificial Intelligence
that enables computers to understand, process, and generate human language.

NLP is used in applications such as chatbots, machine translation, sentiment
analysis, text classification, speech assistants, and information retrieval.

Modern NLP systems often use transformer-based models because they can
capture relationships between words and understand the context of sentences.
        """,
        metadata={
            "source": "NLP Overview",
            "topic": "NLP",
            "page": 1
        }
    ),

    Document(
        page_content="""
Deep Learning is a subfield of Machine Learning that uses artificial neural
networks with multiple layers to learn complex patterns from large amounts
of data.

Deep learning is widely used in image recognition, speech recognition,
natural language processing, recommendation systems, and autonomous
vehicles.

Popular deep learning architectures include Convolutional Neural Networks
(CNNs), Recurrent Neural Networks (RNNs), and Transformer models.
        """,
        metadata={
            "source": "Deep Learning Basics",
            "topic": "DL",
            "page": 1
        }
    )
]
```

```python
vectorstore.add_documents(sample_documents)
```

```python
results = vectorstore.similarity_search(
    "How does a computer learn from data?",
    k=2
)

for doc in results:
    print(doc.metadata)
    print(doc.page_content)
    print("-" * 50)
```
