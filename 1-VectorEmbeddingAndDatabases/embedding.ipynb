### VECTOR DATABASES

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Word_embedding = {
    "cat":[0.8,0.6],
    "kitten":[0.75,0.65],
    "dog":[0.7,0.3],
    "puppy":[0.65,0.35],
    "car":[-0.5,0.2],
    "truck":[-0.45,0.15]
}
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))

for word, coords in Word_embedding.items():
    ax.scatter(coords[0], coords[1], s=100)

    ax.annotate(
        word,
        (coords[0], coords[1]),
        xytext=(5, 5),
        textcoords="offset points"
    )

ax.set_xlabel("Dimension 1")
ax.set_ylabel("Dimension 2")
ax.set_title("Simplified Word Embeddings in 2D Space")
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Measuring Similarity

```python
def cosine_similarity(vect1,vect2):
  """ 
  cosine similarity is measure the angle between two vector
   - result close to 1 :very similar
   - result close to 0 : not related
   - result close to -1 : opposite meaning

  """
  
  dot_product = np.dot(vect1,vect2)
  magnitude_a =np.linalg.norm(vect1)
  magnitude_b =np.linalg.norm(vect2)
  
  return dot_product/(magnitude_a*magnitude_b)
      
```

```python
cat=[0.8,0.6,0.3],
kitten=[0.75,0.65,0.35]

cat_kitten_similarity = cosine_similarity(cat,kitten)
print(cat_kitten_similarity)

```

## Creating our first embedding

```python
## hugging-face and open-ai embedding


from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-miniLM-L6-V2"
)
embeddings
```

```python
#create our first embedding
text = " Hello iam learning about embedding"
embedding = embeddings.embed_query(text)
print(f"Text: {text}")
print(f"Embedding Length {len(embedding)}")
print(embedding)
```
